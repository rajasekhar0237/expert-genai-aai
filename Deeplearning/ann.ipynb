```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:,2 ]= le.fit_transform(X[:,2])
```

```python
X
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
print(X)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state =0)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

ANN MODEL BUILDING ARCHITECHTURE

```python
ann = tf.keras.models.Sequential() # establish layer one after another
```

```python
ann.add(tf.keras.layers.Dense(units=6))# input layer
```

```python
ann.add(tf.keras.layers.Dense(units=6,activation='relu'))## hidden layer1
```

```python
ann.add(tf.keras.layers.Dense(units=1,activation='sigmoid'))# out put layer
```

```python
ann.compile(optimizer='adam',loss ='binary_crossentropy',metrics=['accuracy'])
```

```python
ann.fit(X_train,y_train,batch_size=32,epochs= 50)
```
